```python
import networkx as nx
```

```python
def read_miss_trace_file(f: str, max_count=8192) -> nx.MultiDiGraph:
    gr = nx.DiGraph()
    
    with open(f, 'rb') as rd:
        s_count_bytes = rd.read(4)
        ii = 0
        while s_count_bytes != '' and ii < max_count:
            s_count = int.from_bytes(rd.read(4), byteorder='little')
            installed = int.from_bytes(rd.read(8), byteorder='little')
            victim = int.from_bytes(rd.read(8), byteorder='little')

            if gr.has_edge(installed, victim):
                gr.edges[installed,victim]['weight'] += 1
            else:
                gr.add_edge(installed, victim, weight=1)

            s_count_bytes = rd.read(4)
            ii += 1

    return gr
```
